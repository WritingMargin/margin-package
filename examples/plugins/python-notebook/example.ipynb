# A small experiment in reading time
Run the cells in order to see how notebook variables carry across cells. These results appear only in your editor preview; this source notebook intentionally has no saved outputs.

```python
passage = 'A good article starts a conversation. A notebook lets readers explore an idea.'
words = len(passage.split())
print(f'This passage contains {words} words.')
```

```python
words_per_minute = 220
seconds = words / words_per_minute * 60
print(f'At {words_per_minute} words per minute, that takes about {seconds:.1f} seconds to read.')
```
